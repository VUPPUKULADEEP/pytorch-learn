```python
import torch
```

```python
x = torch.rand(3, requires_grad = True)
x
```

```python
# it have gradient function according to operation
y = x*x*2
y
```

```python
# example of mean function
z = x.mean()
z
```

```python
# we can able to see the gradient vals 
# how much you need to change the weight
# for forward propagation no need of gradients
print(x.grad)
```

```python
# now one time we go back
z.backward()
```

```python
# now we can see the gradients
print(x.grad)
```

```python
# if output is scalar no need to send the arguments to backward
# if output contain vector then we can send our tensor to backward
z = x**2
v = torch.tensor([0.1,0.01,0.001], dtype=torch.float64)
z.backward(v)
print(x.grad)
```

```python
# by using these we can modify the requries_grad= False in run time
# x.requires_grad_(False)
# x.detach() # it creates a other tensor copy from the original
# with torch.no_grad():
#     pass

# by these we can use the tensor with out the gradient
```

```python
# a sample script to demonsrate the epochs

weights = torch.ones(4, requires_grad = True)


for epoch in range(2):
    model_output = (weights*3).sum()
    model_output.backward()
    print(weights.grad)
    weights.grad.zero_()
```

```python
# with optimizer
# demonstrating how the output changes
weights = torch.ones(4, requires_grad = True)
optimizer = torch.optim.SGD([weights], lr = 0.01)
for epoch in range(5):
    model_output = (weights*3).sum()
    print(model_output)
    model_output.backward()
    print(weights.grad)
    optimizer.step()
    print(weights)
    optimizer.zero_grad()
    
```
