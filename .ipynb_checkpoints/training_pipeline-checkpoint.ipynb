```python
import torch
import torch.nn as nn
from termcolor import colored
```

# design model (input, output size, forwardpass,loss, backward, update weight )

```python
x = torch.tensor([[1],[2],[3],[4]], dtype= torch.float32)
y = torch.tensor([[2],[4],[6],[8]], dtype= torch.float32)
```

```python
n_samples, n_features = x.shape
print(n_samples, n_features)
x_test = torch.tensor([5], dtype= torch.float32)
input_size, output_size = n_features, n_features
model = nn.Linear(input_size, output_size)
# w = torch.tensor(0.0, dtype=torch.float32, requires_grad = True)
```

```python
print(f'preciction before training:f(5) = {model(x_test).item():.3f}')

learning_rate = 0.01
n_iters = 100
# loss
# no need manually
loss = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)


for epoch in range(n_iters):
    y_pred = model(x)
    l = loss(y, y_pred)
    
    print(colored(f'epoch {epoch}', "red"))
    print(colored(f'loss {l}', "green"))
    l.backward()
    optimizer.step()
    [w, b] = model.parameters()
    print(colored(f'gradients after backward {w, b}', "blue"))
    optimizer.zero_grad()
    print(colored(f'updated weights  {w}', "yellow"))

print(f'preciction after training:f(5) = {model(x_test).item():.3f}')
    
```
