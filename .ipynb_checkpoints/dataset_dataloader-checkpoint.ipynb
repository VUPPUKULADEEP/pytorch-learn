```python
import torch
import torchvision
from torch.utils.data import Dataset, DataLoader
import numpy as np
import math
```

```python
class WineDataSet(Dataset):
    def __init__(self):
        #data loading
        xy = np.loadtxt('dataset.csv', delimiter = ",", dtype=np.float32, skiprows = 1)
        x = np.concatenate([xy[0:30, 1:],xy[60:90, 1:],xy[130:150, 1:]])
        y = np.concatenate([xy[0:30, [0]],xy[60:90, [0]],xy[130:150, [0]]])
        x = (x - x.mean(axis=0)) / x.std(axis=0)
        self.x = torch.from_numpy(x).float()
        self.y = torch.from_numpy(y).float() #n_samples, 1
        self.n_samples = x.shape[0]
    def __getitem__(self, index):
        #dataset[0]
        return self.x[index], self.y[index]
    def __len__(self):
        #gives the length of dataset
        return self.n_samples
```

```python
# using dataset

dataset = WineDataSet()
first_data = dataset[0]
features, lables = first_data
print(features, lables)
```

```python
# using DataLoader

dataloader = DataLoader(dataset = dataset, batch_size = 1, shuffle= True, num_workers = 0)

dataiter = iter(dataloader)
```

```python
# checking length
dataset.__len__()
```

```python
# training loop

num_epochs = 10
total_samples = len(dataset)
n_iterations = math.ceil(total_samples/4)
print(total_samples, n_iterations)
```

```python
# dummy training loop

for epoch in range(num_epochs):
    for i , (inputs,labels) in enumerate(dataloader):
        if (i+1)%5 == 0:
            print(f'epoch {epoch+1}/{num_epochs} step{i+1}/{n_iterations} inputs {inputs.shape}')
```

```python
# linear regression nn 13 inputs 1 output 13 inputs are features, 1 output as labels

class LinearModel(torch.nn.Module):
    def __init__(self):
        super(LinearModel, self).__init__()
        self.linear = torch.nn.Linear(13, 1)
    def forward(self, x):
        return self.linear(x)
```

```python
model = LinearModel()
loss = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr = 0.01)
```

# linear regression of wine dataset 13 inputs, 1 output

```python
# linear regression training loop

for epoch in range(num_epochs):
    for i , (inputs,labels) in enumerate(dataloader):
        y_hat = model(inputs)
        l = loss(y_hat, labels)
        l.backward()
        optimizer.step()
        optimizer.zero_grad()
        print(f"Epoch [{epoch + 1}/{num_epochs}], Loss: {l.item():.4f}")
```

```python
model.eval()
```

```python
with torch.no_grad():
    prediction = model(dataset.x[0].unsqueeze(0))

print("Actual:", dataset.y[0].item())
print("Predicted:", prediction.item())
```
