```python
import torch
from termcolor import colored
```

```python
w = torch.tensor(1.0, requires_grad = True)
```

```python
x = torch.tensor(1.0)
y = torch.tensor(2.0)
optimizer = torch.optim.SGD([w], lr=0.1)
```

```python
# forward pass

y_hat = w*x
loss = (y_hat - y)**2
print(loss)
```

```python
# backward pass
loss.backward()
print(w.grad)
```

```python
# update n=1 learning rate
optimizer.step()
print(w)
optimizer.zero_grad()
```

```python
# again forward pass
y_hat = w*x
loss = (y_hat - y)**2
print(loss)
```

```python
# again backward
loss.backward()
print(w.grad)
```

```python
# adain weight update

optimizer.step()
print(w)
optimizer.zero_grad()
```

```python
# simply run the epochs in loop

for epoch in range(20):
    print(colored(f'epoch {epoch}', "red"))
    y_hat = w*x
    loss = (y_hat - y)**2
    print(colored(f'loss {loss}', "green"))
    loss.backward()
    print(colored(f'gradients after backward {w.grad}', "blue"))
    optimizer.step()
    print(colored(f'updated weights  {w}', "yellow"))
    optimizer.zero_grad()
```
